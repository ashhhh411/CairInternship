EXAMPLE 1: SIMILAR TWEETS

```python
from sentence_transformers import SentenceTransformer

# Load sentence transformer model
model = SentenceTransformer("all-MiniLM-L6-v2")

print("Model loaded successfully! 🎯")
```

```python
# Sample tweets
tweets = [
    "AI is transforming the world!",
    "Machine learning is amazing.",
    "Deep learning improves AI models.",
    "Big data is essential for AI training.",
    "Natural language processing is a key AI field."
]

# Generate embeddings
tweet_embeddings = model.encode(tweets)

print("Generated embeddings successfully! ✅")
```

```python
from qdrant_client import QdrantClient
from qdrant_client.models import Distance, VectorParams

# Connect to Qdrant server (local or cloud)
client = QdrantClient("localhost", port=6333)  # Update IP if using cloud

print("Connected to Qdrant! 🚀")
```

```python
# Define collection parameters
client.recreate_collection(
    collection_name="tweets_collection",
    vectors_config=VectorParams(size=384, distance=Distance.COSINE)
)

print("Collection created successfully! ✅")
```

```python
# Insert data into Qdrant
client.upsert(
    collection_name="tweets_collection",
    points=[
        {"id": i, "vector": tweet_embeddings[i].tolist(), "payload": {"text": tweets[i]}}
        for i in range(len(tweets))
    ]
)

print("Tweets stored in Qdrant successfully! 🎯")
```

```python
# Encode a query tweet
query = "Artificial Intelligence is changing industries."
query_vector = model.encode(query).tolist()

# Search for similar tweets in Qdrant
search_results = client.search(
    collection_name="tweets_collection",
    query_vector=query_vector,
    limit=3  # Return top 3 most similar tweets
)

# Print results
print("🔍 Top 3 similar tweets:")
for result in search_results:
    print(f"- {result.payload['text']} (score: {result.score})")
```

EXAMPLE 2: SIMILAR TWEETS

```python
tweets = [
    "Building a vector database pipeline for AI-powered search!",
    "Want real-time insights? Use vector search with Qdrant!",
    "AI + Vector Search = Next-gen recommendations",
    "How to store embeddings efficiently? Try Qdrant!",
    "Data-driven decision-making starts with efficient search.",
    "Qdrant makes similarity search blazing fast!",
]
```

```python
from sentence_transformers import SentenceTransformer  

# Load a transformer model for embedding tweets  
model = SentenceTransformer("all-MiniLM-L6-v2")  

# Generate embeddings  
tweet_embeddings = model.encode(tweets)  
```

```python
from qdrant_client import QdrantClient  
from qdrant_client.models import Distance, VectorParams  

# Connect to Qdrant (Localhost)  
client = QdrantClient("localhost", port=6333)  

# Define collection name  
collection_name = "tweets1_collection"

# Create a collection in Qdrant  
client.recreate_collection(  
    collection_name=collection_name,  
    vectors_config=VectorParams(size=384, distance=Distance.COSINE)  
)
```

```python
client.upsert(  
    collection_name=collection_name,  
    points=[  
        {"id": i, "vector": tweet_embeddings[i].tolist(), "payload": {"text": tweets[i]}}  
        for i in range(len(tweets))  
    ]  
)  
```

```python
query_text = "How to make AI-powered search efficient?"  

# Convert query to an embedding  
query_embedding = model.encode([query_text]).tolist()[0]  

# Perform a similarity search in Qdrant  
search_results = client.search(  
    collection_name=collection_name,  
    query_vector=query_embedding,  
    limit=3  # Return top 3 most similar tweets  
)  

# Display results  
for result in search_results:  
    print(f"Tweet: {result.payload['text']} (Score: {result.score:.4f})")  
```
